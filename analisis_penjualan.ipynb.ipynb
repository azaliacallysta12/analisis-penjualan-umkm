```python

import pandas as pd

df = pd.read_csv("data_penjualan.csv")

print("Dataset berhasil dibaca!")
print("Jumlah baris dan kolom:", df.shape)

display(df)
```

```python

# Menghitung jumlah barang dan pendapatan per produk
analisis = df.groupby("Produk").agg(
    Jumlah_Terjual=("Jumlah", "sum"),
    Total_Pendapatan=("Total", "sum")
)

print("HASIL ANALISIS PENJUALAN")
display(analisis)

# Produk dengan jumlah penjualan tertinggi
produk_terlaris = analisis["Jumlah_Terjual"].idxmax()

print("Produk paling banyak terjual:", produk_terlaris)
print("Total pendapatan toko: Rp",
      f"{df['Total'].sum():,}".replace(",", "."))
```

```python

import matplotlib.pyplot as plt

# Grafik jumlah barang terjual
analisis["Jumlah_Terjual"].plot(
    kind="bar",
    title="Jumlah Barang Terjual per Produk",
    xlabel="Nama Produk",
    ylabel="Jumlah Terjual"
)

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Grafik pendapatan setiap produk
analisis["Total_Pendapatan"].plot(
    kind="bar",
    title="Pendapatan per Produk",
    xlabel="Nama Produk",
    ylabel="Pendapatan (Rupiah)"
)

plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

analisis.to_csv("hasil_analisis.csv")

print("Hasil analisis berhasil disimpan!")
```
